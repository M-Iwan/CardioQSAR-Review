```python
max_study = max(len(item) for item in df['Study'].to_list())
max_data = max(len(item) for item in df['Data'].to_list())
max_code = max(len(item) for item in df['Code'].to_list())
max_model = max(len(item) for item in df['Model'].to_list())
max_platform = max(len(item) for item in df['Platform'].to_list())
max_paper = max(len(item) for item in df['In-paper'].to_list())
```

```python
for row in df.iter_rows():
    study, data, code, model, platform, paper = row
    e_study    = " " * (max_study    - len(study))
    e_data     = " " * (max_data     - len(data))
    e_code     = " " * (max_code     - len(code))
    e_model    = " " * (max_model    - len(model))
    e_platform = " " * (max_platform - len(platform))
    e_paper    = " " * (max_paper    - len(paper))

    print(f"\t {study}{e_study} & {data}{e_data} & {code}{e_code} & {model}{e_model} & {platform}{e_platform} & {paper}{e_paper} &  \\\\")
```
